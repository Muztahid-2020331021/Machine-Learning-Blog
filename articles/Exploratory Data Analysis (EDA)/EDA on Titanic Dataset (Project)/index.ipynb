At first we need to import the necessary libraries 

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

Now we will load the dataset. You can download the dataset from <a href="train.csv" download>here</a>.

```python
data = pd.read_csv('train.csv')
data.head()
```

```python
data.shape
```

```python
data.info()
```

```python
data.isnull().sum()
```

```python
sns.countplot(x='Survived', data=data)
```

```python
sns.countplot(x='Pclass', data=data)
```

**Insights:**
If we look at the above graph, we can see that most of the passengers who were from the 3rd class in the Titanic.


If we want to do the multivariate analysis, we can do like this. 

```python
sns.countplot(data = data, x = 'Survived', hue='Pclass')
```

**Insights:**
If we look at the above graph, we can see that most of the passengers who died were from the 3rd class. This is because the 3rd class had less access to lifeboats and safety measures compared to the 1st and 2nd classes. 
On the other hand, most of the passengers who survived were from the 1st class. This is because the 1st class had better access to lifeboats and safety measures compared to the 2nd and 3rd classes.

```python
sns.countplot(data = data, x = 'Survived', hue='Sex')
```

**Insights:**
If we look at the above graph, we can see that most of the passengers who died were male. On the other hand, most of the survivors were female.

```python
sns.displot(data=data, x='Age', kde = True)
```

**Insights:**
There are some outliers in the age column. We can see that the graph is skewed to the right, which means that there are more younger passengers than older passengers. The majority of the passengers were between the ages of 20 and 40.

## Feature Engineering

Feature engineering is the process of using domain knowledge to extract features from raw data. These features can be used to improve the performance of machine learning algorithms.

At first, we do Exploratory Data Analysis (EDA) to understand the data and identify the problem we are trying to solve. These problems are solved in the Feature Engineering step. 

The features are also known as columns or independent variables. 

There is a concept called "GIGO" which stands for "Garbage In, Garbage Out". That means if we have a poor model and we feed it with good data, we will get good results. But if we have a good model and we feed it with poor data, we will get poor results.

So, the steps are:  `Data Loading -> EDA -> Feature Engineering -> Model Training`

Let's try to understand what are in the feature engineering.

```python
Feature Engineering
├── Feature Transformation
│   ├── Missing Value Handling
│   ├── Handling Categorical Feature
│   ├── Outlier Detection
│   └── Feature Scaling
├── Feature Construction
├── Feature Selection
└── Feature Extraction
```

Here is a high-level overview of the feature engineering process.

### Feature Transformation

1. **Missing Value Handling**:

    Here is an image for example. 

    ![Missing Value Handling.png](images/im1.png)

    In the image, we can see that some cells values are `NaN` which means this dataset contains missing values. The most popular way to handle missing values is to fill them with the mean. In the image, we can fill the missing values with the mean that is 26.0. Sometimes there are outliers in the dataset. In that case, we can fill the missing values with the median. The median is less affected by outliers than the mean. We will see them in practical later.

2. **Handling Categorical Feature**:

    Here is an image for example.

    ![Handling Categorical Feature.png](images/im2.png)

    We can not use categorical features directly in machine learning algorithms. We need to convert them into numerical features. The most popular way to handle categorical features is to use one-hot encoding. The image shows the one-hot encoding. There are other ways to handle categorical data. We will dicuss more later.

3. **Outlier Detection**:

    Here is an image for example.

    ![Outlier Detection.png](images/im3.png)

    In the image, we can see that there are some outliers in the left graph. Thats why the linear regression line is not fitting well. After removing the outliers(right graph), the linear regression line is fitting well. So they must be removed by the IQR method for better performance of some models.

4. **Feature Scaling**:

    ![Feature Scaling.png](images/im4.png)

    If we look at this data, we can see that the value range of the `Age` column is very lower than the value range of the `Salary` column. So the dataset is not scaled. If we use this dataset to train the models that works on distance, the model will give more importance to the `Salary` column than the `Age` column. So we need to scale the dataset which is not expected. There is another problem with gradient descent(which is used for reducing the loss). If we use this dataset, the gradient descent will not converge because of the different value ranges. So we need to scale the dataset. 



### Feature Construction

We can create new features from the existing features by feature construction. 

![Feature Construction.png](images/im5.png)

Look at this titanic dataset. We can see that there are two features 'SibSp' and 'Parch'. Suppose we want to create a new feature 'Family' which is the sum of 'SibSp' and 'Parch'. By using this new feature, we can create a new feature 'FamilySize' which is a categorical feature. If 'Family' == 1, then 'FamilySize' = 'Alone', if 'Family' > 1 and 'Family' <= 4, then 'FamilySize' = 'Medium', if 'Family' > 4, then 'FamilySize' = 'Large'. This is an example of feature construction. This helps us to reduce the curse of dimensionality.

### Feature Selection

We will use MNIST handwritten digit dataset for this example.

![Feature Selection1.png](images/im6.png)

![Feature Selection2.png](images/im7.png)

In the image, we can see that there are so many features we don't need. We can select the most important features by using feature selection and omit other features. This helps us to reduce the curse of dimensionality and improve the performance of the model.

### Feature Extraction

Suppose we have two independent features `Number of Rooms` and `Number of Bedrooms` and we have a target variable `Price`. We have to predict the price of a house based on the number of rooms and bedrooms. Our target is to remove the two features and create a new feature `Area`. There may be a confusion between feature construction and feature extraction. The difference is that in feature construction, we create new features from the existing features by calculation, but in feature extraction, we try to maintain the meaning of the removed features with the new feature.
